# Data Understanding
Analyzing the new raw source files replacing the old synthetic datasets.

```python
import pandas as pd
import os

# Load all primary raw datasets
df_main = pd.read_csv('../data/raw/employee_performance_pro.csv')
df_perf = pd.read_csv('../data/raw/Employee_Performance_Dataset.csv')
df_wa   = pd.read_csv('../data/raw/WA_Fn-UseC_-HR-Employee-Attrition.csv')
df_hr   = pd.read_csv('../data/raw/Cleaned_HR_Data_Analysis.csv')

datasets = {
    'employee_performance_pro.csv': df_main,
    'Employee_Performance_Dataset.csv': df_perf,
    'WA_Fn-UseC_-HR-Employee-Attrition.csv': df_wa,
    'Cleaned_HR_Data_Analysis.csv': df_hr
}

for name, df in datasets.items():
    print(f'--- {name} ---')
    print(f'Shape: {df.shape}')
    print(f'Columns: {list(df.columns)}')
    print(f'Missing Values (top 5):\n{df.isnull().sum().sort_values(ascending=False).head(5)}')
    print(f'Duplicates: {df.duplicated().sum()}')
    print()
```

```python
# Check Attrition target class balance in our primary dataset
print('Attrition Class Balance (employee_performance_pro.csv):')
print(df_main['AttritionRisk'].value_counts(normalize=True).round(4))
print()
print('Attrition Class Balance (IBM WA Dataset):')
print(df_wa['Attrition'].value_counts(normalize=True).round(4))
```

### Dataset Summary
**employee_performance_pro.csv**: Primary HR table. Key = `EmployeeID`. Contains the ML target `AttritionRisk` (Yes/No), demographic info, salary, work-life balance, and performance rating.

**Employee_Performance_Dataset.csv**: Extended KPI performance metrics. Key = `Employee ID`. Contains `KPI Score`, `Peer Rating`, `Task Completion (%)`, and `Promotion Eligibility`.

**Cleaned_HR_Data_Analysis.csv**: Organizational/engagement data. Key = `Employee ID`. Contains `Engagement Score`, `Satisfaction Score`, `Business Unit`, and training information.

**WA_Fn-UseC_-HR-Employee-Attrition.csv**: Classic IBM Attrition dataset (1470 rows). Key = `EmployeeNumber`. Useful for supplemental ML training due to rich feature set (MonthlyIncome, JobSatisfaction, etc.).

**Join Strategy**: `employee_performance_pro.csv` is the spine. Merge `Employee_Performance_Dataset.csv` and `Cleaned_HR_Data_Analysis.csv` on `Employee ID` → `EmployeeID`.
