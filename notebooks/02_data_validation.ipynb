# Data Validation
Validates the new real datasets for schema conformance, missing values, and distribution checks.

```python
import pandas as pd
import numpy as np

df_main = pd.read_csv('../data/raw/employee_performance_pro.csv')
df_perf = pd.read_csv('../data/raw/Employee_Performance_Dataset.csv')
df_hr   = pd.read_csv('../data/raw/Cleaned_HR_Data_Analysis.csv')

# --- Schema validation ---
required_main = ['EmployeeID', 'Department', 'JobRole', 'Age', 'MonthlySalary', 'YearsAtCompany', 'AttritionRisk', 'PerformanceRating', 'WorkLifeBalanceScore']
required_perf = ['Employee ID', 'Performance Score', 'KPI Score', 'Attendance (%)', 'Peer Rating', 'Task Completion (%)', 'Promotion Eligibility']
required_hr   = ['Employee ID', 'Engagement Score', 'Satisfaction Score']

for name, df, req in [('employee_performance_pro', df_main, required_main), ('Employee_Performance_Dataset', df_perf, required_perf), ('Cleaned_HR_Data_Analysis', df_hr, required_hr)]:
    missing = [c for c in req if c not in df.columns]
    print(f'[{name}] Shape: {df.shape}')
    if missing:
        print(f'  WARNING: Missing expected columns: {missing}')
    else:
        print(f'  OK: All required columns present.')
    print(f'  Nulls: {df.isnull().sum().sum()} total null values')
    print(f'  Duplicates: {df.duplicated().sum()}')
    print()
```

```python
# --- Distribution checks ---
print('AttritionRisk distribution:')
print(df_main['AttritionRisk'].value_counts())
print()
print('Age distribution (min/max/mean):', df_main['Age'].min(), '/', df_main['Age'].max(), '/', round(df_main['Age'].mean(), 1))
print('MonthlySalary (min/max/mean):', df_main['MonthlySalary'].min(), '/', df_main['MonthlySalary'].max(), '/', round(df_main['MonthlySalary'].mean(), 0))
print('YearsAtCompany (min/max):', df_main['YearsAtCompany'].min(), '/', df_main['YearsAtCompany'].max())
print()
print('Department distribution:')
print(df_main['Department'].value_counts())
```
