# Engagement Intelligence
Analyzing engagement and satisfaction scores (sourced from HR dataset dept-level aggregates).

```python
import pandas as pd
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv('../data/processed/employee_attrition_processed.csv')
print(f'Loaded: {df.shape}')

eng_cols = [c for c in ['avg_EngagementScore', 'avg_SatisfactionScore', 'avg_WLB_Score', 'WorkLifeBalance'] if c in df.columns]
print(f'Engagement columns: {eng_cols}')
print(f'Missing in any: {df[eng_cols].isnull().sum().sum()}')
print()
print('Engagement metrics by Attrition status:')
print(df.groupby('Attrition_encoded')[eng_cols].mean().round(2))
```

```python
# Plot engagement distribution by attrition status
col = eng_cols[0] if eng_cols else None
if col:
    retained = df[df['Attrition_encoded'] == 0][col].dropna()
    at_risk = df[df['Attrition_encoded'] == 1][col].dropna()
    
    if len(retained) > 0 and len(at_risk) > 0 and retained.std() > 0 and at_risk.std() > 0:
        fig, ax = plt.subplots(figsize=(10, 5))
        ax.hist(retained, bins=15, alpha=0.7, color='#38bdf8', label=f'Retained (n={len(retained)})')
        ax.hist(at_risk, bins=15, alpha=0.7, color='#ef4444', label=f'At Risk (n={len(at_risk)})')
        ax.set_title(f'{col} by Attrition Status')
        ax.set_xlabel(col)
        ax.legend()
        plt.tight_layout()
        plt.savefig('../data/processed/engagement_distribution.png', dpi=100, bbox_inches='tight')
        plt.close()
        print('Chart saved.')
    else:
        print(f'Skipping chart - insufficient variance in {col} (dept-level aggregates have low variance by design).')

# Show department-level engagement breakdown
print('\nEngagement by Department:')
print(df.groupby('Department')['avg_EngagementScore'].mean().sort_values(ascending=False).round(2) if 'avg_EngagementScore' in df.columns else 'No EngagementScore available')
```
