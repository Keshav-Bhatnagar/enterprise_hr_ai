# Data Relationships
Exploring correlations and join quality between the 3 real data sources.

```python
import pandas as pd

df_main = pd.read_csv('../data/raw/employee_performance_pro.csv')
df_perf = pd.read_csv('../data/raw/Employee_Performance_Dataset.csv')
df_hr   = pd.read_csv('../data/raw/Cleaned_HR_Data_Analysis.csv')

# Check join coverage: how many employees in main match the performance dataset?
main_ids = set(df_main['EmployeeID'])
perf_ids = set(df_perf['Employee ID'])
hr_ids   = set(df_hr['Employee ID'])

print(f'Main dataset employees: {len(main_ids)}')
print(f'Performance dataset employees: {len(perf_ids)}')
print(f'HR/Org dataset employees: {len(hr_ids)}')
print()
print(f'Main <-> Performance join coverage: {len(main_ids & perf_ids)} / {len(main_ids)} ({100*len(main_ids & perf_ids)/len(main_ids):.1f}%)')
print(f'Main <-> HR/Org join coverage: {len(main_ids & hr_ids)} / {len(main_ids)} ({100*len(main_ids & hr_ids)/len(main_ids):.1f}%)')
```

```python
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv('../data/processed/employee_attrition_processed.csv')

# Correlation with Attrition_encoded target
num_df = df.select_dtypes(include='number')
corr = num_df.corr()['Attrition_encoded'].drop('Attrition_encoded').abs().sort_values(ascending=False)
print('Top 10 features correlated with Attrition Risk:')
print(corr.head(10))
```
